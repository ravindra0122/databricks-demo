## What is Databricks
Databricks is a unified data analytics platform built on top of Apache Spark. It provides managed clusters, collaborative notebooks, job scheduling, and integrations with data storage and ML tools.

## Typical workflow
1. Ingest data into cloud storage (S3, ADLS, or DBFS).
2. Explore and transform data with notebooks (PySpark/SQL).
3. Train models or run analytics jobs.
4. Schedule jobs and productionize pipelines.

```python
# Minimal PySpark example (works on Databricks or any PySpark environment)
from pyspark.sql import SparkSession

# On Databricks, a SparkSession is usually provided as `spark` already.
spark = SparkSession.builder.appName("databricks-intro").getOrCreate()

data = [(1, 'Alice'), (2, 'Bob'), (3, 'Charlie')]
columns = ['id', 'name']
df = spark.createDataFrame(data, columns)

print('Schema:')
df.printSchema()

print('Data:')
df.show()
```
